# Argumento para la distinción didáctica

**El argumento se construye contrastando dos niveles de desempeño sobre estos mismos datos:**

### 1. Ejecución correcta sin comprensión

Un estudiante puede **calcular sin problemas** la media de precipitación:

$$\bar{x} = \frac{0+2+0+5+0+28+7}{7} = \frac{42}{7} = 6 \text{ mm/día}$$

O incluso la mediana (2 mm) y la moda (0 mm). La operación aritmética es impecable. Sin embargo, **el resultado es engañoso como descripción del fenómeno**: "6 mm/día" o "0 mm típico" no capturan que la lluvia se concentró en dos días (Sáb: 28 mm, Dom: 7 mm) y que tres de siete días no llovió. El estudiante ejecutó la operación, pero no comprendió que la media es sensible a valores extremos ni que la distribución es fuertemente asimétrica.

### 2. Comprensión del fenómeno (misma operación, distinta lectura)

Un estudiante que comprende el fenómeno **no se queda en el número**: reconoce que la media (6 mm) está inflada por el valor atípico del sábado, que la mediana (2) refleja correctamente que la mayoría de días fueron de lluvia baja, y que la **desviación estándar** (~10.08 mm) evidencia una dispersión enorme. Puede incluso decidir que la media **no es** la medida adecuada para resumir estos datos, o reportarla junto con el máximo y el número de días secos.

### 3. La distinción queda así demostrada

- **Mismo dataset, misma operación aritmética** → dos estudiantes pueden obtener *exactamente el mismo número* (6 mm).
- **Diferencia didáctica fundamental**: uno *sabe calcular*, el otro *sabe interpretar*. El primero cumple el procedimiento; el segundo comprende qué significa ese número para el fenómeno de la precipitación semanal.
- La tabla es especialmente útil como **ejemplo didáctico** porque el contraste es visible: la media (6) y la mediana (2) discrepan drásticamente, lo que obliga al estudiante a preguntarse *qué mide cada una* y *qué quiere decir "típico"* en este contexto.

**Conclusión:** el script no solo produce números; produce un caso donde *ejecutar correctamente* (calcular la media) y *comprender el fenómeno* (saber que esa media oculta la concentración de lluvia en el fin de semana) son claramente separables. Esa separación es, precisamente, la distinción didáctica fundamental que se quiere ilustrar.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Datos
datos = pd.DataFrame({
    "dia": ["Lun", "Mar", "Mie", "Jue", "Vie", "Sab", "Dom"],
    "precipitacion": [0, 2, 0, 5, 0, 28, 7]
})

# Configuración de la figura
fig, ax = plt.subplots(figsize=(7, 5))

# Boxplot — se usa 'orientation' en lugar de 'vert' (deprecado desde Matplotlib 3.11)
bp = ax.boxplot(
    datos["precipitacion"],
    orientation="vertical",          # <-- reemplaza a vert=True
    patch_artist=True,
    widths=0.5,
    boxprops=dict(facecolor="steelblue", alpha=0.7, edgecolor="black"),
    medianprops=dict(color="red", linewidth=2),
    whiskerprops=dict(color="black"),
    capprops=dict(color="black"),
    flierprops=dict(marker="o", markerfacecolor="orange", markeredgecolor="black",
                    markersize=9)
)

# Etiquetas y título
ax.set_ylabel("Precipitación (mm)")
ax.set_title("Diagrama de cajas y bigotes — precipitación semanal")
ax.set_xticks([1])
ax.set_xticklabels(["Precipitación"])
ax.grid(axis="y", linestyle="--", alpha=0.4)

# Anotar los cinco números del resumen (opcional, muy útil didácticamente)
q1, mediana, q3 = np.percentile(datos["precipitacion"], [25, 50, 75])
iqr = q3 - q1
bigote_inf = max(datos["precipitacion"].min(), q1 - 1.5 * iqr)
bigote_sup = min(datos["precipitacion"].max(), q3 + 1.5 * iqr)

ax.text(1.3, q1,  f"Q1 = {q1:.1f}", va="center", fontsize=9, color="darkblue")
ax.text(1.3, mediana, f"Mediana = {mediana:.1f}", va="center", fontsize=9, color="red")
ax.text(1.3, q3,  f"Q3 = {q3:.1f}", va="center", fontsize=9, color="darkblue")
ax.text(1.3, bigote_sup, f"Bigote sup. = {bigote_sup:.1f}", va="center", fontsize=9)
ax.text(1.3, bigote_inf, f"Bigote inf. = {bigote_inf:.1f}", va="center", fontsize=9)

plt.tight_layout()
plt.savefig('boxplot_precipitacion.png')
plt.show()

# Resumen numérico en consola
print("Resumen de los cinco números:")
print(f"  Mínimo        : {datos['precipitacion'].min():.1f}")
print(f"  Q1 (25%)      : {q1:.1f}")
print(f"  Mediana (50%) : {mediana:.1f}")
print(f"  Q3 (75%)      : {q3:.1f}")
print(f"  Máximo        : {datos['precipitacion'].max():.1f}")
print(f"  IQR           : {iqr:.1f}")
print(f"  Límite inferior (Q1 - 1.5·IQR): {q1 - 1.5*iqr:.1f}")
print(f"  Límite superior (Q3 + 1.5·IQR): {q3 + 1.5*iqr:.1f}")
print(f"  Valores atípicos: {[x for x in datos['precipitacion'] if x < q1-1.5*iqr or x > q3+1.5*iqr]}")
```

### **Un ejemplo de progresión conceptual**

Imagina que los estudiantes investigan la relación entre **horas de estudio** y **resultados académicos**.

Primero aprenden a representar los datos mediante un diagrama de dispersión.

Después estudian la correlación y la regresión lineal.

Posteriormente, en bachillerato, pueden explorar cómo un algoritmo aprende una relación a partir de observaciones y cómo se evalúa una predicción con datos que no se utilizaron para entrenarlo.

```python

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

# 1. Generar datos sintéticos
np.random.seed(42)

horas_estudio = np.random.uniform(0, 10, 50)

# Relación lineal con ruido aleatorio
puntajes = (
    2.5 * horas_estudio
    + 4.5
    + np.random.normal(0, 2, 50)
)

# 2. Preparar los datos para Scikit-learn
X = horas_estudio.reshape(-1, 1)
y = puntajes

# 3. Entrenar el modelo de regresión lineal
modelo = LinearRegression()
modelo.fit(X, y)

# 4. Calcular las predicciones
y_predicho = modelo.predict(X)

# 5. Ordenar los puntos para dibujar la recta
orden = np.argsort(horas_estudio)

# ============================================================
# 6. Construir la secuencia gráfica con plt.subplots()
# ============================================================
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# ------------------------------------------------------------
# ETAPA 1: Solo el diagrama de dispersión
# (Educación primaria / primeros pasos: representar los datos)
# ------------------------------------------------------------
ax1 = axes[0]
ax1.scatter(
    horas_estudio,
    puntajes,
    color="blue",
    label="Puntos de datos",
    s=45,
    alpha=0.8
)
ax1.set_title("Etapa 1: Representar los datos\n(Diagrama de dispersión)",
              fontsize=13, fontweight="bold")
ax1.set_xlabel("Horas estudiadas", fontsize=11)
ax1.set_ylabel("Puntaje en el examen", fontsize=11)
ax1.set_xlim(0, 10)
ax1.set_ylim(3, 30)
ax1.grid(True, alpha=0.6)
ax1.legend(fontsize=10)

# ------------------------------------------------------------
# ETAPA 2: Correlación y regresión lineal
# (Secundaria: estudiar la relación entre variables)
# ------------------------------------------------------------
ax2 = axes[1]
ax2.scatter(
    horas_estudio,
    puntajes,
    color="blue",
    label="Puntos de datos",
    s=45,
    alpha=0.8
)
ax2.plot(
    horas_estudio[orden],
    y_predicho[orden],
    color="red",
    linewidth=2.5,
    label="Regresión lineal"
)
ax2.set_title("Etapa 2: Correlación y regresión\n(La recta que mejor se ajusta)",
              fontsize=13, fontweight="bold")
ax2.set_xlabel("Horas estudiadas", fontsize=11)
ax2.set_ylabel("Puntaje en el examen", fontsize=11)
ax2.set_xlim(0, 10)
ax2.set_ylim(3, 30)
ax2.grid(True, alpha=0.6)
ax2.legend(fontsize=10)

# ------------------------------------------------------------
# ETAPA 3: Entrenamiento y evaluación con datos nuevos
# (Bachillerato: aprendizaje automático y evaluación)
# ------------------------------------------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

modelo_train = LinearRegression()
modelo_train.fit(X_train, y_train)

y_train_pred = modelo_train.predict(X_train)
y_test_pred = modelo_train.predict(X_test)

r2_test = r2_score(y_test, y_test_pred)

ax3 = axes[2]

# Datos de entrenamiento
ax3.scatter(
    X_train.flatten(),
    y_train,
    color="blue",
    label="Entrenamiento",
    s=45,
    alpha=0.8
)

# Datos de prueba (nuevos, no vistos por el modelo)
ax3.scatter(
    X_test.flatten(),
    y_test,
    color="green",
    marker="^",
    s=70,
    label="Prueba (no vistos)",
    edgecolors="black"
)

# Recta aprendida
X_line = np.linspace(0, 10, 100).reshape(-1, 1)
y_line = modelo_train.predict(X_line)
ax3.plot(
    X_line.flatten(),
    y_line,
    color="red",
    linewidth=2.5,
    label="Modelo aprendido"
)

ax3.set_title(
    f"Etapa 3: Aprendizaje y evaluación\nR² en datos nuevos = {r2_test:.3f}",
    fontsize=13,
    fontweight="bold"
)
ax3.set_xlabel("Horas estudiadas", fontsize=11)
ax3.set_ylabel("Puntaje en el examen", fontsize=11)
ax3.set_xlim(0, 10)
ax3.set_ylim(3, 30)
ax3.grid(True, alpha=0.6)
ax3.legend(fontsize=10)

# ------------------------------------------------------------
# Título general y ajuste final
# ------------------------------------------------------------
fig.suptitle(
    "Progresión conceptual: de representar datos a evaluar un modelo",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()

# ============================================================
# 7. Mostrar ecuaciones y métricas
# ============================================================
print("=" * 55)
print("MODELO COMPLETO (todos los datos):")
print(f"y = {modelo.coef_[0]:.3f}x + {modelo.intercept_:.3f}")
print(f"Coeficiente R²: {modelo.score(X, y):.4f}")

print("\nMODELO ENTRENADO (solo con datos de entrenamiento):")
print(f"y = {modelo_train.coef_[0]:.3f}x + {modelo_train.intercept_:.3f}")
print(f"R² entrenamiento: {r2_score(y_train, y_train_pred):.4f}")
print(f"R² prueba (datos nuevos): {r2_test:.4f}")
print(f"RMSE prueba: {np.sqrt(mean_squared_error(y_test, y_test_pred)):.3f}")
print("=" * 55)
```

1. Nivel básico: calcular la media
El estudiante suma las precipitaciones y divide entre siete:
$$ \bar{x}=\frac{0+2+0+5+0+28+7}{7}=6 $$
La precipitación media diaria es de 6 mm por día.
En Python:


```python
import numpy as np
precipitacion = np.array([0, 2, 0, 5, 0, 28, 7])
media = np.mean(precipitacion)
print(f"Precipitación media diaria: {media:.1f} mm")
```

Resultado:   
Precipitación media diaria: 6.0 mm  
El estudiante ha calculado correctamente. Sin embargo, todavía no ha explicado qué sucedió durante esa semana.
2. Nivel interpretativo: utilizar la media para comprender el fenómeno

Ahora formula a tus estudiantes una pregunta diferente:
¿Significa que en Caucasia llovieron 6 mm cada día?  
No. 

La media no describe lo que ocurrió en cada día. Hubo tres días sin lluvia, mientras que el sábado se registraron 28 mm.

Podemos utilizar Python para calcular otros indicadores que ayuden a interpretar el fenómeno:


```python
import numpy as np
precipitacion = np.array([0, 2, 0, 5, 0, 28, 7])
print("Precipitación total:", np.sum(precipitacion), "mm")
print("Media:", np.mean(precipitacion), "mm")
print("Mediana:", np.median(precipitacion), "mm")
print("Máxima precipitación:", np.max(precipitacion), "mm")
print("Días sin lluvia:", np.sum(precipitacion == 0))
print("Días con lluvia:", np.sum(precipitacion > 0))
```

En esta semana ficticia hubo 3 días sin lluvia y 4 días con lluvia.  

Observa que la media es de 6 mm, pero la mediana es de 2 mm.   

La precipitación del sábado influye considerablemente en la media y hace que esta sea superior a la mediana.

Aquí aparece una primera lección de educación estadística: dos medidas de tendencia central pueden aportar perspectivas distintas sobre un mismo conjunto de datos.

```python
import numpy as np
import matplotlib.pyplot as plt

# Datos de precipitación (mm)
precipitacion_semana_A = np.array([0, 2, 0, 5, 0, 28, 7])
precipitacion_semana_B = np.array([6, 6, 6, 6, 6, 6, 6])

# Días de la semana
dias = ["Lun", "Mar", "Mié", "Jue", "Vie", "Sáb", "Dom"]

# ---- Estadísticas por consola ----
for nombre, datos in [("Semana A", precipitacion_semana_A),
                      ("Semana B", precipitacion_semana_B)]:
    print(f"\n{nombre}")
    print("Media:", np.mean(datos))
    print("Mediana:", np.median(datos))
    print("Desviación estándar:", np.std(datos))
    print("Máximo diario:", np.max(datos))

# ---- Gráfico de barras comparativo ----
x = np.arange(len(dias))
ancho = 0.35

fig, ax = plt.subplots(figsize=(10, 6))

barras_A = ax.bar(x - ancho/2, precipitacion_semana_A, ancho,
                  label="Semana A", color="#1f77b4", edgecolor="black")
barras_B = ax.bar(x + ancho/2, precipitacion_semana_B, ancho,
                  label="Semana B", color="#ff7f0e", edgecolor="black")

# Etiquetas de valor sobre cada barra
for barra in list(barras_A) + list(barras_B):
    altura = barra.get_height()
    ax.annotate(f"{altura:.0f}",
                xy=(barra.get_x() + barra.get_width()/2, altura),
                xytext=(0, 3), textcoords="offset points",
                ha="center", va="bottom", fontsize=9)

# Líneas de referencia: medias
ax.axhline(np.mean(precipitacion_semana_A), color="#1f77b4",
           linestyle="--", linewidth=1, alpha=0.7,
           label=f"Media A = {np.mean(precipitacion_semana_A):.2f} mm")
ax.axhline(np.mean(precipitacion_semana_B), color="#ff7f0e",
           linestyle="--", linewidth=1, alpha=0.7,
           label=f"Media B = {np.mean(precipitacion_semana_B):.2f} mm")

ax.set_xlabel("Día de la semana")
ax.set_ylabel("Precipitación (mm)")
ax.set_title("Comparación de precipitaciones: Semana A vs Semana B")
ax.set_xticks(x)
ax.set_xticklabels(dias)
ax.legend(loc="upper left")
ax.grid(axis="y", linestyle=":", alpha=0.6)

plt.tight_layout()
plt.savefig("precipitaciones.png", dpi=150)
plt.show()

# ---- Gráfico de líneas (opcional, evolución diaria) ----
fig2, ax2 = plt.subplots(figsize=(10, 5))
ax2.plot(dias, precipitacion_semana_A, marker="o", linewidth=2,
         color="#1f77b4", label="Semana A")
ax2.plot(dias, precipitacion_semana_B, marker="s", linewidth=2,
         color="#ff7f0e", label="Semana B")
ax2.set_xlabel("Día de la semana")
ax2.set_ylabel("Precipitación (mm)")
ax2.set_title("Evolución diaria de las precipitaciones")
ax2.legend()
ax2.grid(linestyle=":", alpha=0.6)

plt.tight_layout()
plt.savefig("precipitaciones_lineas.png", dpi=150)
plt.show()
```

Pensamiento computacional	Aplicación a la estadística
Descomposición	Dividir una investigación en preguntas, variables, recolección, análisis e interpretación.
Reconocimiento de patrones	Identificar tendencias, diferencias, asociaciones y comportamientos inusuales.
Abstracción	Seleccionar las variables relevantes para responder una pregunta.
Diseño de algoritmos	Definir procedimientos para calcular medidas y generar gráficos.
Automatización	Analizar conjuntos de datos más grandes mediante Python.
Evaluación crítica	Verificar resultados, identificar errores y cuestionar las conclusiones de los modelos.
La idea es que la programación permita hacer más visible el razonamiento matemático. Por ejemplo, al implementar un algoritmo para calcular una media, los estudiantes deben comprender qué representa, cómo se calcula y por qué puede verse afectada por valores extremos.
6. El componente territorial: una oportunidad para diferenciar tu ponencia
Por tu trabajo docente en el Campus Caucasia de la Universidad de Antioquia, puedes darle a la ponencia una perspectiva territorial que vaya más allá de los ejemplos escolares habituales.

En bachillerato, algunas podrían evolucionar hacia el análisis de series de tiempo o la construcción de modelos predictivos, siempre que se disponga de datos pertinentes y suficientes.
La propuesta pedagógica es clara: enseñar estadística investigando el territorio en el que viven los estudiantes. De esta manera, los datos dejan de ser números abstractos y se convierten en herramientas para comprender y transformar su realidad.
7. ¿Qué deberías defender como conclusión?
Te sugiero cerrar con tres ideas fundamentales:
1. La estadística debe enseñarse desde la investigación y no solamente desde el cálculo. Los estudiantes necesitan formular preguntas, recoger información, interpretar representaciones y argumentar con evidencias.
2. El pensamiento computacional puede fortalecer la comprensión estadística. Python facilita la experimentación, la visualización y la automatización, pero su valor educativo depende de la comprensión matemática que acompaña su uso.
3. La alfabetización estadística es una condición para comprender críticamente la inteligencia artificial. Antes de confiar en las predicciones de un modelo, es necesario comprender la calidad de los datos, la incertidumbre, los errores, los sesgos y los límites de las inferencias.
Mi recomendación es que no presentes el Machine Learning y el Deep Learning como contenidos que necesariamente deban enseñarse formalmente en primaria. Preséntalos como el horizonte al que puede conducir una formación estadística bien construida, respetando el desarrollo cognitivo de los estudiantes y la progresión curricular.

8. Resumen para proponer la ponencia al simposio
Título: De las tablas de frecuencia a la inteligencia artificial: enseñar estadística para desarrollar el pensamiento computacional en primaria y bachillerato.
Resumen
La enseñanza de la estadística en la educación primaria y secundaria enfrenta el desafío de trascender el aprendizaje mecánico de procedimientos y fórmulas para promover la interpretación crítica de datos, la formulación de preguntas y la argumentación basada en evidencias. En este contexto, el desarrollo del pensamiento computacional y el uso educativo de herramientas de programación ofrecen oportunidades para transformar las prácticas de enseñanza y aprendizaje de la estadística.
Esta ponencia propone una perspectiva didáctica que articula el pensamiento estadístico, el pensamiento computacional y la alfabetización en ciencia de datos e inteligencia artificial. Se plantea una progresión formativa que comienza con la recolección, organización y representación de datos en situaciones cercanas a los estudiantes; continúa con la exploración de medidas estadísticas, patrones y relaciones entre variables; e incorpora, de manera gradual, la programación con Python como herramienta para automatizar cálculos, visualizar información y explorar modelos predictivos sencillos.
Mediante ejemplos de aplicación y una demostración con herramientas como NumPy, Pandas y Matplotlib, se examina cómo los problemas del entorno escolar y territorial pueden convertirse en experiencias de investigación estadística. Asimismo, se discute la importancia de comprender la variabilidad, la incertidumbre, los sesgos y las limitaciones de los datos para interpretar responsablemente las predicciones producidas por modelos de Machine Learning y Deep Learning.
Se concluye que la formación estadística puede constituirse en un fundamento para desarrollar el pensamiento crítico, la competencia digital y la capacidad de comprender fenómenos reales, siempre que la tecnología se integre al servicio de la comprensión matemática y de una enseñanza contextualizada.
